```python
# Cell 26: Install and import transformers for FinBERT
# Run this in a separate cell first if needed:
# !pip install transformers torch

import torch

from transformers import pipeline

# Initialize FinBERT
print("Loading FinBERT model...")
finbert = pipeline("sentiment-analysis", 
                   model="ProsusAI/finbert",
                   device=0 if torch.cuda.is_available() else -1)
print("FinBERT loaded!")
```

```python
# Cell 27: Apply sentiment analysis in batches
import numpy as np
from tqdm import tqdm

# Function to get sentiment scores
def get_sentiment_scores(texts, batch_size=32):
    sentiments = []
    
    for i in tqdm(range(0, len(texts), batch_size)):
        batch = texts[i:i+batch_size].tolist()
        # Get predictions
        results = finbert(batch)
        sentiments.extend(results)
    
    return sentiments

# Sample first to test (use smaller sample for testing)
print("Testing on 100 headlines first...")
sample_headlines = df_filtered['headline'].head(100)
sample_sentiments = get_sentiment_scores(sample_headlines, batch_size=16)

print(f"\nSample results:")
for i in range(3):
    print(f"Headline: {sample_headlines.iloc[i][:80]}...")
    print(f"Sentiment: {sample_sentiments[i]}")
    print()
```

```python
# Cell 28: Process all headlines - THIS WILL TAKE A WHILE
# Only run this after confirming the sample works

print(f"Processing {len(df_filtered)} headlines...")
print("This will take approximately 10-15 minutes...")

# Process in chunks to avoid memory issues
df_filtered['sentiment_label'] = None
df_filtered['sentiment_score'] = None

# Process
all_sentiments = get_sentiment_scores(df_filtered['headline'], batch_size=32)

# Extract labels and scores
df_filtered['sentiment_label'] = [s['label'] for s in all_sentiments]
df_filtered['sentiment_score'] = [s['score'] for s in all_sentiments]

print("Sentiment analysis complete!")
print("\nSentiment distribution:")
print(df_filtered['sentiment_label'].value_counts())
```

```python
# Cell 29: Convert sentiments to numerical values
# Map sentiment labels to numbers
sentiment_map = {
    'positive': 1,
    'neutral': 0,
    'negative': -1
}

df_filtered['sentiment_numeric'] = df_filtered['sentiment_label'].map(sentiment_map)

# Create weighted sentiment (direction * confidence)
df_filtered['sentiment_weighted'] = df_filtered['sentiment_numeric'] * df_filtered['sentiment_score']

print("Sentiment encoding complete!")
print("\nSample of processed data:")
print(df_filtered[['date', 'stock', 'headline', 'sentiment_label', 'sentiment_weighted']].head())
```

```python
# Cell 30: Aggregate sentiment by date and stock
# Group by date and stock, calculate daily sentiment metrics
daily_sentiment = df_filtered.groupby(['date', 'stock']).agg({
    'sentiment_weighted': ['mean', 'sum', 'std'],
    'sentiment_numeric': ['mean', 'count'],
    'headline': 'count'
}).reset_index()

# Flatten column names
daily_sentiment.columns = ['date', 'stock', 'sentiment_mean', 'sentiment_sum', 
                          'sentiment_std', 'sentiment_raw_mean', 'sentiment_count', 'news_count']

# Fill NaN std with 0 (when only 1 news item)
daily_sentiment['sentiment_std'].fillna(0, inplace=True)

print("Daily aggregated sentiment:")
print(daily_sentiment.head())
print(f"\nShape: {daily_sentiment.shape}")
```

```python
# Cell 31 - FIXED V2: Properly structure price data
all_price_data = []

for stock, prices in price_data.items():
    # Create a clean dataframe for this stock
    df_stock = pd.DataFrame()
    df_stock['date'] = pd.to_datetime(prices.index)
    df_stock['stock'] = stock
    df_stock['Close'] = prices['Close'].values
    df_stock['Volume'] = prices['Volume'].values
    
    # Calculate returns and direction
    df_stock['returns'] = df_stock['Close'].pct_change()
    df_stock['direction'] = 0  # Initialize
    df_stock.loc[df_stock['returns'] > 0, 'direction'] = 1
    df_stock.loc[df_stock['returns'] < 0, 'direction'] = -1
    df_stock['direction_binary'] = (df_stock['returns'] > 0).astype(int)
    
    all_price_data.append(df_stock)

# Combine properly
df_prices = pd.concat(all_price_data, ignore_index=True)

print(f"Price data shape: {df_prices.shape}")
print(f"Columns: {df_prices.columns.tolist()}")
print("\nPrice data sample (should have NO NaN except first return):")
print(df_prices.head(10))

# Verify no weird structure
print(f"\nData check:")
print(f"Unique stocks: {df_prices['stock'].unique()}")
print(f"Any NaN in Close?: {df_prices['Close'].isna().sum()}")
print(f"Any NaN in Volume?: {df_prices['Volume'].isna().sum()}")
```

```python
# Cell 33: Merge sentiment and price data
df_final = pd.merge(
    daily_sentiment,
    df_prices,
    on=['date', 'stock'],
    how='inner'
)

# Sort by stock and date
df_final.sort_values(['stock', 'date'], inplace=True)

# Drop first row of each stock (no returns)
df_final = df_final.dropna(subset=['returns'])

# Create target: predict NEXT day's direction
df_final['next_day_direction'] = df_final.groupby('stock')['direction_binary'].shift(-1)

# Drop last row of each stock (no next day)
df_final = df_final.dropna(subset=['next_day_direction'])

# Convert target to int
df_final['next_day_direction'] = df_final['next_day_direction'].astype(int)

print(f"Final dataset shape: {df_final.shape}")
print(f"Columns: {df_final.columns.tolist()}")
print(f"\nTarget distribution:")
print(f"Up days: {(df_final['next_day_direction']==1).sum()} ({(df_final['next_day_direction']==1).mean():.1%})")
print(f"Down days: {(df_final['next_day_direction']==0).sum()} ({(df_final['next_day_direction']==0).mean():.1%})")
```

```python
# Cell 34: Select and scale features
from sklearn.preprocessing import StandardScaler

# Select features for LSTM
feature_columns = ['sentiment_mean', 'sentiment_sum', 'sentiment_std', 
                   'news_count', 'Close', 'Volume', 'returns']

# Create feature matrix
X = df_final[feature_columns].values
y = df_final['next_day_direction'].values

print(f"Features shape: {X.shape}")
print(f"Target shape: {y.shape}")
print(f"\nFeatures being used:")
for i, col in enumerate(feature_columns):
    print(f"{i}: {col}")

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(f"\nScaled features sample:")
print(X_scaled[:3])
```

```python
# Cell 35: Create sequences for LSTM
import numpy as np

def create_sequences(X, y, sequence_length=10):
    """
    Create sequences for LSTM
    sequence_length: number of previous days to use for prediction
    """
    X_seq, y_seq = [], []
    
    for i in range(sequence_length, len(X)):
        X_seq.append(X[i-sequence_length:i])
        y_seq.append(y[i])
    
    return np.array(X_seq), np.array(y_seq)

# Create sequences (using 10 days to predict next day)
SEQUENCE_LENGTH = 10
X_sequences, y_sequences = create_sequences(X_scaled, y, SEQUENCE_LENGTH)

print(f"Sequences shape: {X_sequences.shape}")
print(f"Targets shape: {y_sequences.shape}")
print(f"\nInterpretation:")
print(f"- {X_sequences.shape[0]} samples")
print(f"- {X_sequences.shape[1]} days of history per sample")
print(f"- {X_sequences.shape[2]} features per day")
```

```python
# Cell 36: Train/validation/test split
from sklearn.model_selection import train_test_split

# First split: separate test set (80/20)
X_temp, X_test, y_temp, y_test = train_test_split(
    X_sequences, y_sequences, test_size=0.2, random_state=42, shuffle=False
)

# Second split: separate train and validation (80/20 of remaining)
X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=0.2, random_state=42, shuffle=False
)

print(f"Training set: {X_train.shape}")
print(f"Validation set: {X_val.shape}")
print(f"Test set: {X_test.shape}")
print(f"\nClass distribution:")
print(f"Train - Up: {y_train.sum()}/{len(y_train)} ({y_train.mean():.1%})")
print(f"Val - Up: {y_val.sum()}/{len(y_val)} ({y_val.mean():.1%})")
print(f"Test - Up: {y_test.sum()}/{len(y_test)} ({y_test.mean():.1%})")
```

```python
# Cell 37: Import PyTorch and build LSTM model
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

# Check if GPU is available
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Define LSTM model
class StockLSTM(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers, dropout=0.2):
        super(StockLSTM, self).__init__()
        
        self.hidden_size = hidden_size
        self.num_layers = num_layers
        
        # LSTM layers
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True,
            dropout=dropout if num_layers > 1 else 0
        )
        
        # Fully connected layers
        self.fc1 = nn.Linear(hidden_size, 16)
        self.dropout = nn.Dropout(dropout)
        self.fc2 = nn.Linear(16, 1)
        self.relu = nn.ReLU()
        self.sigmoid = nn.Sigmoid()
        
    def forward(self, x):
        # LSTM forward pass
        lstm_out, _ = self.lstm(x)
        
        # Take the last output
        last_output = lstm_out[:, -1, :]
        
        # Pass through fully connected layers
        out = self.fc1(last_output)
        out = self.relu(out)
        out = self.dropout(out)
        out = self.fc2(out)
        out = self.sigmoid(out)
        
        return out

# Initialize model
input_size = X_train.shape[2]  # Number of features
hidden_size = 50
num_layers = 2
dropout = 0.2

model = StockLSTM(input_size, hidden_size, num_layers, dropout).to(device)
print(f"Model structure:\n{model}")
```

```python
# Cell 38: Convert data to PyTorch tensors and create DataLoaders
# Convert numpy arrays to PyTorch tensors
X_train_tensor = torch.FloatTensor(X_train)
y_train_tensor = torch.FloatTensor(y_train).unsqueeze(1)

X_val_tensor = torch.FloatTensor(X_val)
y_val_tensor = torch.FloatTensor(y_val).unsqueeze(1)

X_test_tensor = torch.FloatTensor(X_test)
y_test_tensor = torch.FloatTensor(y_test).unsqueeze(1)

# Create datasets
train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
val_dataset = TensorDataset(X_val_tensor, y_val_tensor)
test_dataset = TensorDataset(X_test_tensor, y_test_tensor)

# Create data loaders
batch_size = 32

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

print(f"Train batches: {len(train_loader)}")
print(f"Val batches: {len(val_loader)}")
print(f"Test batches: {len(test_loader)}")
```

```python
# Cell 39: Define training and evaluation functions
def train_epoch(model, data_loader, criterion, optimizer, device):
    model.train()
    total_loss = 0
    correct = 0
    total = 0
    
    for X_batch, y_batch in data_loader:
        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)
        
        # Forward pass
        outputs = model(X_batch)
        loss = criterion(outputs, y_batch)
        
        # Backward pass
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        # Calculate accuracy
        predicted = (outputs > 0.5).float()
        total += y_batch.size(0)
        correct += (predicted == y_batch).sum().item()
        total_loss += loss.item()
    
    avg_loss = total_loss / len(data_loader)
    accuracy = 100 * correct / total
    
    return avg_loss, accuracy

def evaluate(model, data_loader, criterion, device):
    model.eval()
    total_loss = 0
    correct = 0
    total = 0
    
    with torch.no_grad():
        for X_batch, y_batch in data_loader:
            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)
            
            outputs = model(X_batch)
            loss = criterion(outputs, y_batch)
            
            predicted = (outputs > 0.5).float()
            total += y_batch.size(0)
            correct += (predicted == y_batch).sum().item()
            total_loss += loss.item()
    
    avg_loss = total_loss / len(data_loader)
    accuracy = 100 * correct / total
    
    return avg_loss, accuracy
```

```python
# Cell 40: Training loop
# Setup training parameters
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, patience=5, factor=0.5)

# Training loop
num_epochs = 50
best_val_acc = 0
patience = 10
patience_counter = 0

train_losses = []
val_losses = []
train_accs = []
val_accs = []

print("Starting training...")
for epoch in range(num_epochs):
    # Train
    train_loss, train_acc = train_epoch(model, train_loader, criterion, optimizer, device)
    
    # Validate
    val_loss, val_acc = evaluate(model, val_loader, criterion, device)
    
    # Store metrics
    train_losses.append(train_loss)
    val_losses.append(val_loss)
    train_accs.append(train_acc)
    val_accs.append(val_acc)
    
    # Learning rate scheduling
    scheduler.step(val_loss)
    
    # Print progress
    if (epoch + 1) % 5 == 0:
        print(f"Epoch [{epoch+1}/{num_epochs}]")
        print(f"  Train Loss: {train_loss:.4f}, Train Acc: {train_acc:.2f}%")
        print(f"  Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.2f}%")
    
    # Early stopping
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), 'best_model.pth')
        patience_counter = 0
    else:
        patience_counter += 1
        
    if patience_counter >= patience:
        print(f"Early stopping at epoch {epoch+1}")
        break

print(f"\nBest validation accuracy: {best_val_acc:.2f}%")
```

```python
# Cell 41: Re-run sentiment with full probabilities
from transformers import pipeline
import numpy as np
from tqdm import tqdm

# Initialize FinBERT with full scores
finbert = pipeline(
    "sentiment-analysis",
    model="ProsusAI/finbert",
    device=-1,  # CPU
    return_all_scores=True,  # GET ALL PROBABILITIES
    truncation=True
)

def get_full_sentiment(texts, batch_size=32):
    """Get full probability distribution for each text"""
    all_results = []
    
    for i in tqdm(range(0, len(texts), batch_size)):
        batch = texts[i:i+batch_size].tolist()
        results = finbert(batch, truncation=True, max_length=512)
        all_results.extend(results)
    
    return all_results

# Test on small sample first
print("Testing on 100 headlines...")
sample_results = get_full_sentiment(df_filtered['headline'].head(100))

# Extract probabilities
def extract_probs(result):
    probs = {r['label'].lower(): r['score'] for r in result}
    return probs['positive'], probs['neutral'], probs['negative']

# Show sample
for i in range(3):
    pos, neu, neg = extract_probs(sample_results[i])
    print(f"Headline: {df_filtered['headline'].iloc[i][:60]}...")
    print(f"Positive: {pos:.3f}, Neutral: {neu:.3f}, Negative: {neg:.3f}")
    print(f"Sentiment Score: {pos - neg:.3f}\n")
```
